## Example showing the computation of distances from the coordinate file

#### Output is a matrix of distances between the residues (Mg atom for chl and center of mass for CARs)

```python
import os.path
import numpy as np
import matplotlib.pyplot as plt
import math
```

--------------------------

```python
## importing function from src folder
```

```python
from src import get_distance as GD
```

```python
target_coord_file = 'resd_coord_file_sel.DAT'
```

------------------------

```python
# use the predefined function which processes the DAT file
```

```python
chl_res_ids, chl_coords, car_res_ids, car_coords = GD.extract_distance(target_coord_file)
```

```python
# following code block computes the distance between the the two points in 3D-space
```

```python
nCar = car_coords.shape[0]
nChl = chl_coords.shape[0]

dm = np.zeros((nCar, nChl))

for i in range(nCar):
    for j in range(nChl):
        
        coord1 = car_coords[i,:]
        coord2 = chl_coords[j,:]
        distance = math.sqrt( (coord2[0] - coord1[0])**2 + \
                             (coord2[1] - coord1[1])**2+ \
                             (coord2[2] - coord1[2])**2 ) 
        dm [i,j] = distance
```
